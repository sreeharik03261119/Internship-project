# Barchart

```python
category_col = "Crop_Category"   # Replace with your categorical column
value_col = "Rainfall_mm"                 # Replace with your numerical column

# Calculate average value by category
avg_data = df.groupby(category_col)[value_col].mean().sort_values(ascending=False)

# Create bar chart
plt.figure(figsize=(10,6))
bars = plt.bar(avg_data.index, avg_data.values)

# Add values on top of bars
for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f"{bar.get_height():.1f}",
        ha='center',
        va='bottom',
        fontsize=10
    )

plt.title(f"Average {value_col} by {category_col}", fontsize=16)
plt.xlabel(category_col, fontsize=12)
plt.ylabel(f"Average {value_col}", fontsize=12)
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()

plt.show()
```

# Detect Outliers

```python
import matplotlib.pyplot as plt

num_cols = df.select_dtypes(include=['int64','float64']).columns

for col in num_cols:
    plt.figure(figsize=(5,3))
    plt.boxplot(df[col])
    plt.title(col)
    plt.show()
```

```python
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower) & (df[col] <= upper)]
```
