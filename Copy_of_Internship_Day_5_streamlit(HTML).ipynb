```python
%%writefile app.py
import os
import pandas as pd
from flask import Flask, render_template, request
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.cluster import KMeans
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

app = Flask(__name__)

# -----------------------------------------------------------------
# Load data + train model ONCE at startup
# -----------------------------------------------------------------
CSV_PATH = os.path.join(os.path.dirname(__file__), "agricultural_supply_chain_20_columns.csv")

raw = pd.read_csv(CSV_PATH)
raw = raw.drop(columns=["Record_ID"]).dropna()

df_enc = raw.copy()
cat_cols = df_enc.select_dtypes(include="object").columns.tolist()

encoders = {}
for col in cat_cols:
    enc = LabelEncoder()
    df_enc[col] = enc.fit_transform(df_enc[col].astype(str))
    encoders[col] = enc

target_encoder = encoders["Crop_Name"]

X_clean = df_enc.drop(columns=["Crop_Name"])
y_clean = df_enc["Crop_Name"]

cluster_scaler = StandardScaler()
X_scaled = cluster_scaler.fit_transform(X_clean)

cluster_model = KMeans(n_clusters=3, random_state=42, n_init=10)
X_clean = X_clean.copy()
X_clean["Cluster"] = cluster_model.fit_predict(X_scaled)

X_train, X_test, y_train, y_test = train_test_split(
    X_clean, y_clean, test_size=0.2, random_state=42
)

model = DecisionTreeClassifier(criterion="gini", random_state=42)
model.fit(X_train, y_train)
ACCURACY = round(model.score(X_test, y_test) * 100, 2)

# -----------------------------------------------------------------
# Manual input fields
# -----------------------------------------------------------------
MANUAL_FIELDS = [
    "Month", "Quarter", "Continent", "Country",
    "Climate_Zone", "Currency", "Farming_System", "Crop_Category",
    "Commodity_Type", "Crop_Growth_Stage",
    "Farm_Size_ha", "Soil_Type", "Soil_pH", "Soil_Moisture_pct", "Rainfall_mm",
]

FIELD_OPTIONS = {
    col: list(encoders[col].classes_) for col in MANUAL_FIELDS if col in encoders
}

# Meaningful grouping - mirrors how an agronomist would actually
# think about the inputs: where/when, what conditions, what crop/market.
FIELD_GROUPS = [
    ("Section A", "Time & place", ["Month", "Quarter", "Continent", "Country", "Climate_Zone", "Currency"]),
    ("Section B", "Field conditions", ["Farming_System", "Soil_Type", "Crop_Growth_Stage",
                                        "Farm_Size_ha", "Soil_pH", "Soil_Moisture_pct", "Rainfall_mm"]),
    ("Section C", "Crop & market", ["Crop_Category", "Commodity_Type"]),
]


def build_form_groups():
    """Return fields grouped into tagged sections, in the shape the template needs."""
    groups = []
    for tag, group_name, cols in FIELD_GROUPS:
        fields = []
        for col in cols:
            if col in FIELD_OPTIONS:
                fields.append({"name": col, "label": col.replace("_", " "), "type": "select",
                                "options": FIELD_OPTIONS[col]})
            else:
                fields.append({"name": col, "label": col.replace("_", " "), "type": "number"})
        groups.append({"tag": tag, "name": group_name, "fields": fields})
    return groups


MODEL_STATS = {
    "algorithm": "Decision Tree (Gini)",
    "features": len(X_clean.columns),
    "train_rows": len(X_train),
    "test_rows": len(X_test),
    "classes": len(target_encoder.classes_),
}


@app.route("/", methods=["GET"])
def index():
    return render_template(
        "index.html",
        accuracy=ACCURACY,
        stats=MODEL_STATS,
        groups=build_form_groups(),
        prediction=None,
    )


@app.route("/predict", methods=["POST"])
def predict_route():
    row = {}
    for col in MANUAL_FIELDS:
        if col in encoders:
            row[col] = encoders[col].transform([request.form[col]])[0]
        else:
            row[col] = float(request.form[col])

    row["Date"] = encoders["Date"].transform([raw["Date"].mode().iloc[0]])[0]
    row["Year"] = float(raw["Year"].mean())
    row["Crop_Variety"] = encoders["Crop_Variety"].transform(
        [raw["Crop_Variety"].mode().iloc[0]]
    )[0]

    feature_cols = X_clean.drop(columns=["Cluster"]).columns
    base_row = pd.DataFrame([row])[feature_cols]
    row["Cluster"] = int(cluster_model.predict(cluster_scaler.transform(base_row))[0])

    final_input = pd.DataFrame([row])[X_clean.columns]
    prediction = model.predict(final_input)
    predicted_crop = target_encoder.inverse_transform(prediction)[0]
    confidence = round(float(model.predict_proba(final_input).max()) * 100, 1)

    return render_template(
        "index.html",
        accuracy=ACCURACY,
        stats=MODEL_STATS,
        groups=build_form_groups(),
        prediction=predicted_crop,
        confidence=confidence,
        selected=request.form,
    )


if __name__ == "__main__":
    app.run(host="0.0.0.0", port=8501, debug=False)
```
